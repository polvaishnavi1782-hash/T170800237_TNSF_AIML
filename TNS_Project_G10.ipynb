```python
# Separating features and target variable

X = df.drop('heart_disease', axis=1)
y = df['heart_disease']

print("Features (X) shape:", X.shape)
print("Target (y) shape:", y.shape)
```

```python
# Feature Scaling

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Feature scaling completed successfully!")
print("Scaled data shape:", X_scaled.shape)
```

```python
# Train-Test Split

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Splitting the original features and target
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Scaling the features
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

```python
# Logistic Regression Model

from sklearn.linear_model import LogisticRegression

# Creating the model
logistic_model = LogisticRegression(random_state=42)

# Training the model
logistic_model.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")
```

```python
# Decision Tree Model

from sklearn.tree import DecisionTreeClassifier

# Creating the model
decision_tree_model = DecisionTreeClassifier(
    random_state=42
)

# Training the model
decision_tree_model.fit(X_train, y_train)

print("Decision Tree model trained successfully!")
```

```python
# Random Forest Model

from sklearn.ensemble import RandomForestClassifier

# Creating the model
random_forest_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Training the model
random_forest_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python
# Support Vector Machine (SVM) Model

from sklearn.svm import SVC

# Creating the model
svm_model = SVC(
    probability=True,
    random_state=42
)

# Training the model
svm_model.fit(X_train, y_train)

print("SVM model trained successfully!")
```

```python
# Making predictions on the test data

y_pred_lr = logistic_model.predict(X_test)
y_pred_dt = decision_tree_model.predict(X_test)
y_pred_rf = random_forest_model.predict(X_test)
y_pred_svm = svm_model.predict(X_test)

print("Predictions generated successfully!")
```

```python
# Model Evaluation - Accuracy, Precision, Recall and F1-score

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

models_predictions = {
    "Logistic Regression": y_pred_lr,
    "Decision Tree": y_pred_dt,
    "Random Forest": y_pred_rf,
    "SVM": y_pred_svm
}

results = []

for model_name, predictions in models_predictions.items():
    accuracy = accuracy_score(y_test, predictions)
    precision = precision_score(y_test, predictions)
    recall = recall_score(y_test, predictions)
    f1 = f1_score(y_test, predictions)

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    })

results_df = pd.DataFrame(results)

results_df
```

```python
# Calculating Specificity for all models

from sklearn.metrics import confusion_matrix

specificity_results = []

for model_name, predictions in models_predictions.items():

    tn, fp, fn, tp = confusion_matrix(
        y_test,
        predictions
    ).ravel()

    specificity = tn / (tn + fp)

    specificity_results.append({
        "Model": model_name,
        "Specificity": specificity
    })

specificity_df = pd.DataFrame(specificity_results)

specificity_df
```

```python
# Confusion Matrices for all models

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

for ax, (model_name, predictions) in zip(
    axes.ravel(),
    models_predictions.items()
):

    cm = confusion_matrix(y_test, predictions)

    sns.heatmap(
        cm,
        annot=True,
        fmt='d',
        cmap='Blues',
        cbar=False,
        ax=ax
    )

    ax.set_title(model_name)
    ax.set_xlabel('Predicted')
    ax.set_ylabel('Actual')

plt.suptitle('Confusion Matrices of Classification Models', fontsize=16)

plt.tight_layout()

plt.show()
```

```python
# ROC-AUC Score for all models

from sklearn.metrics import roc_auc_score

roc_auc_results = []

models = {
    "Logistic Regression": logistic_model,
    "Decision Tree": decision_tree_model,
    "Random Forest": random_forest_model,
    "SVM": svm_model
}

for model_name, model in models.items():

    # Probability of the positive class (heart disease = 1)
    y_prob = model.predict_proba(X_test)[:, 1]

    # Calculate ROC-AUC
    auc_score = roc_auc_score(y_test, y_prob)

    roc_auc_results.append({
        "Model": model_name,
        "ROC-AUC": auc_score
    })

roc_auc_df = pd.DataFrame(roc_auc_results)

roc_auc_df
```

```python
# Combining all evaluation metrics

final_results = results_df.merge(
    specificity_df,
    on="Model"
).merge(
    roc_auc_df,
    on="Model"
)

# Display values rounded to 4 decimal places
final_results = final_results.round(4)

final_results
```

```python
# ROC Curves for all models

from sklearn.metrics import roc_curve

plt.figure(figsize=(9, 7))

for model_name, model in models.items():

    # Get probability of heart disease = 1
    y_prob = model.predict_proba(X_test)[:, 1]

    # Calculate ROC curve
    fpr, tpr, thresholds = roc_curve(y_test, y_prob)

    # Calculate AUC
    auc_score = roc_auc_score(y_test, y_prob)

    # Plot ROC curve
    plt.plot(
        fpr,
        tpr,
        label=f'{model_name} (AUC = {auc_score:.3f})'
    )

# Random classifier reference line
plt.plot(
    [0, 1],
    [0, 1],
    linestyle='--',
    label='Random Classifier'
)

plt.title('ROC Curves for Classification Models')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')

plt.legend()
plt.grid()

plt.show()
```

```python
# Identifying the best model based on F1-Score

best_model_name = final_results.loc[
    final_results['F1-Score'].idxmax(),
    'Model'
]

best_f1_score = final_results['F1-Score'].max()

print("Best Model:", best_model_name)
print("Best F1-Score:", best_f1_score)
```

```python
# Performance summary of the best model

best_model_results = final_results[
    final_results['Model'] == best_model_name
]

best_model_results
```

```python
# Visual comparison of all classification models

comparison_metrics = [
    'Accuracy',
    'Precision',
    'Recall',
    'F1-Score',
    'Specificity',
    'ROC-AUC'
]

final_results.set_index('Model')[comparison_metrics].plot(
    kind='bar',
    figsize=(12, 7)
)

plt.title('Comparison of Classification Models')
plt.xlabel('Models')
plt.ylabel('Score')
plt.ylim(0, 1)

plt.xticks(rotation=0)
plt.legend(title='Metrics')

plt.tight_layout()
plt.show()
```

```python
# Creating the FastAPI application

from fastapi import FastAPI
import numpy as np

app = FastAPI(
    title="Heart Disease Detection API",
    description="API for predicting heart disease using a trained classification model",
    version="1.0"
)

print("FastAPI application created successfully!")
```

```python
# Creating the prediction endpoint

from pydantic import BaseModel

class PatientData(BaseModel):
    age: int
    sex: int
    chest_pain_type: int
    resting_blood_pressure: int
    cholesterol: int
    fasting_blood_sugar: int
    resting_ecg: int
    max_heart_rate: int
    exercise_induced_angina: int
    st_depression: float
    st_slope: int
    num_major_vessels: int
    thalassemia: int


@app.post("/predict")
def predict_heart_disease(patient: PatientData):

    # Convert patient data into an array
    input_data = np.array([[
        patient.age,
        patient.sex,
        patient.chest_pain_type,
        patient.resting_blood_pressure,
        patient.cholesterol,
        patient.fasting_blood_sugar,
        patient.resting_ecg,
        patient.max_heart_rate,
        patient.exercise_induced_angina,
        patient.st_depression,
        patient.st_slope,
        patient.num_major_vessels,
        patient.thalassemia
    ]])

    # Scale the input
    input_scaled = loaded_scaler.transform(input_data)

    # Make prediction
    prediction = loaded_model.predict(input_scaled)[0]

    # Return result
    if prediction == 1:
        result = "Heart Disease Detected"
    else:
        result = "No Heart Disease Detected"

    return {
        "prediction": int(prediction),
        "result": result
    }


print("Prediction endpoint created successfully!")
```
